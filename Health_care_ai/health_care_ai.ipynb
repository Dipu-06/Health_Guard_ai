```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier,VotingClassifier
from sklearn.metrics import accuracy_score,classification_report,recall_score
from sklearn.model_selection import train_test_split

!pip install xgboost
```

```python
df=pd.read_csv("novagen_dataset.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
df.dtypes
```

```python
x=df.drop("Target",axis=1)
y=df["Target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python

from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

<h1>LOGISTIC REGRESSION</h1>

```python
model1=LogisticRegression(max_iter=1000,penalty='l2',random_state=42,solver="liblinear")
model1.fit(x_train_scaled,y_train)
y_pred=model1.predict(x_test_scaled)
print(f"accuracy{accuracy_score(y_pred,y_test)}")
print(f"recall{recall_score(y_pred,y_test)}")
print(f"clasification_report{classification_report(y_pred,y_test)}")

```

<h1>KNN</h1>

```python
model2 = KNeighborsClassifier(n_neighbors=5,metric="euclidean")
model2.fit(x_train_scaled,y_train)
y_pred=model2.predict(x_test_scaled)
print(f"accuracy {accuracy_score(y_pred,y_test)}")
print(f"recall {recall_score(y_pred,y_test)}")
print(classification_report(y_pred,y_test))
```

<h1>Random forest classifier</h1>

```python
model3=RandomForestClassifier(n_estimators=200)
model3.fit(x_train_scaled,y_train)
y_pred=model3.predict(x_test_scaled)
print(f"accuracy {accuracy_score(y_pred,y_test)}")
print(f"recall {recall_score(y_pred,y_test)}")
print(classification_report(y_pred,y_test))
```

<h1>GradientBoosting<h1>

```python
model4 = GradientBoostingClassifier(n_estimators=1000, learning_rate=1.0,
    max_depth=1, random_state=0).fit(x_train_scaled, y_train)
y_pred=model4.predict(x_test_scaled)
print(f"accuracy {accuracy_score(y_pred,y_test)}")
print(f"recall {recall_score(y_pred,y_test)}")
print(classification_report(y_pred,y_test))
```

<h1>vootingclassifier<h1>

```python
model5=VotingClassifier(estimators=[
    ("lr",LogisticRegression()),
    ("knn",KNeighborsClassifier()),
    ("randomforest",RandomForestClassifier())
],voting='hard')
model5.fit(x_train_scaled,y_train)
y_pred=model5.predict(x_test_scaled)
print(f"accuracy {accuracy_score(y_pred,y_test)}")
print(f"recall {recall_score(y_pred,y_test)}")
print(classification_report(y_pred,y_test))
```
